> **Nota sobre uso de Programação Orientada a Objetos (POO):**  
> Nos Exercícios 9 e 10, o uso de Programação Orientada a Objetos (POO) não é obrigatório, mas é fortemente recomendado. Independentemente do paradigma adotado, o aluno precisa resolver explicitamente o problema de persistência de estado da estrutura de dados. Pilhas e filas mantêm informações internas que evoluem ao longo do tempo (array de armazenamento, índices, tamanho, capacidade), e essas informações não podem ser recriadas a cada operação. O uso de classes facilita essa persistência ao encapsular estado e comportamento em uma única unidade, tornando o código mais claro, modular e previsível. Caso o aluno opte por uma abordagem não orientada a objetos, deverá demonstrar de forma igualmente clara e correta como o estado da pilha ou fila é mantido e atualizado entre as operações, sem perda de informação.

### Exercício 12: Análise de Funções Hash e Distribuição

**Introdução**  
A eficiência de uma hash table depende diretamente da qualidade da função hash. Funções mal distribuídas geram colisões e degradam o desempenho das operações. Neste exercício, você irá analisar criticamente diferentes funções hash simples, avaliando seu impacto na distribuição das chaves.

**Funções hash fornecidas**  
Considere uma tabela hash de tamanho fixo $M = 10$ e as seguintes funções hash para chaves inteiras não negativas $k$:
- $h_1(k) = k \bmod M$
- $h_2(k) = (k \times 7) \bmod M$
- $h_3(k) = (k^2) \bmod M$

**Tarefa**  
1. Analise as funções hash fornecidas considerando uma tabela de tamanho fixo.
2. Avalie a distribuição gerada para diferentes conjuntos de chaves.
3. Identifique as funções com maior risco de colisão.
4. Justifique suas conclusões tecnicamente.
5. Registre a análise em Markdown.

**Observações**  
• Não implemente uma hash table completa. O foco do exercício é a qualidade da função hash e sua distribuição.  
• A clareza da análise faz parte da avaliação.
